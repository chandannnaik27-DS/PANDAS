## Deffernt ways of creating Data Frames

### Using CSV

```python
import pandas as pd

pd.read_csv('weather_data.csv')
```

### Using excel

```python
import pandas as pd

df = pd.read_excel("Weather_data.xlsx", sheet_name="Sheet1")

print(df)
```

### Using Python Dictonary

```python
# Make one python dectionary of required data set and then read it as a data frame 

weather_data = {
    'day' : ['1/4/2026', '2/4/2026', '3/4/2026', '4/4/2026'],
    'temperature' : [32, 28, 34, 35],
    'climate' : ['Sunny','snow','sunny','sunny']
}
pd.DataFrame(weather_data)
```

### Using Python Tuples list

```python
weather_data = [
    ('1/4/2026',32,6,'Sunny'),
    ('2/4/2026',28,4,'snow'),
    ('3/4/2026',34,5,'sunny'),
    ('4/4/2026',35,3,'sunny')
]
pd.DataFrame(weather_data, columns=["day","tempertaure","windspeed","climate"])
```

### Using list of dictionaries

```python
weather_data = [
    {'day':'1/4/2026', 'temperature':32, 'windspeed':6, 'climate':'sunny'},
    {'day':'2/4/2026', 'temperature':28, 'windspeed':4, 'climate':'snow'},
    {'day':'3/4/2026', 'temperature':34, 'windspeed':5, 'climate':'sunny'},
    {'day':'4/4/2026', 'temperature':35, 'windspeed':3, 'climate':'sunny'}
]
pd.DataFrame(weather_data)
```
