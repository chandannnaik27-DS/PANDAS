### Reading excel file

```python
import pandas as pd

pd.read_excel("stock_data(excel).xlsx")
```

```python

def convert_people_cell(cell):
    if cell == "n.a.":
        return "Sam walton"
    return cell

def convert_eps_cell(cell):
    if cell == "not available":
        return None
    return cell
pd.read_excel("stock_data(excel).xlsx",converters={
    'people' : convert_people_cell,
    'eps' : convert_eps_cell
})
```
